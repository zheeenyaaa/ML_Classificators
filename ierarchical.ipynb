```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

# 1. Генерация тестовых данных
X, y = make_blobs(n_samples=50, centers=3, random_state=42)

# 2. Важный шаг: стандартизация признаков
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 3. Построение матрицы связи через SciPy для дендрограммы
linked = linkage(X_scaled, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(linked)
plt.title('Дендрограмма иерархической кластеризации')
plt.xlabel('Индекс образца')
plt.ylabel('Расстояние')
plt.show()

# 4. Применение AgglomerativeClustering из scikit-learn
from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
labels = cluster.fit_predict(X_scaled)

print('Метки кластеров:', labels)
```
